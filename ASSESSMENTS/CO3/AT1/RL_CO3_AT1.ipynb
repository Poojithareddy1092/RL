```python
#Agent–Environment Interaction: States, Actions and Rewards
# Agent-Environment Interaction

states = [0, 1, 2, 3, 4]
actions = ["left", "right"]

position = 0
total_reward = 0

for step in range(10):

    # Agent chooses an action
    if position < 4:
        action = "right"
    else:
        action = "left"

    # Environment responds
    if action == "right":
        position += 1
    else:
        position -= 1

    # Reward
    if position == 4:
        reward = 10
    else:
        reward = -1

    total_reward += reward

    print("State:", position,
          "Action:", action,
          "Reward:", reward)

print("Cumulative Reward:", total_reward)
```

```python
#2. Modeling Decision-Making as an MDP
# Simple Markov Decision Process

states = ["A", "B", "C"]
actions = ["left", "right"]

# Transition probabilities
transition = {
    "A": {"right": "B", "left": "A"},
    "B": {"right": "C", "left": "A"},
    "C": {"right": "C", "left": "B"}
}

# Reward function
rewards = {
    ("A", "right"): 1,
    ("A", "left"): -1,
    ("B", "right"): 10,
    ("B", "left"): 0,
    ("C", "right"): 0,
    ("C", "left"): -1
}

gamma = 0.9       # Discount factor

state = "A"
total_reward = 0

for i in range(5):

    action = "right"

    next_state = transition[state][action]
    reward = rewards[(state, action)]

    total_reward = total_reward + reward

    print("State:", state,
          "Action:", action,
          "Next State:", next_state,
          "Reward:", reward)

    state = next_state

print("Total Reward:", total_reward)
print("Discount Factor:", gamma)
```

```python
#3. Policy, Value Functions and Bellman Equation
# Value Functions using Bellman Equation

states = ["A", "B", "C"]

# Initial state values
V = {
    "A": 0,
    "B": 0,
    "C": 10
}

# Rewards
R = {
    ("A", "right"): 2,
    ("B", "right"): 5,
    ("C", "right"): 0
}

# Next states
next_state = {
    ("A", "right"): "B",
    ("B", "right"): "C",
    ("C", "right"): "C"
}

gamma = 0.9

# Bellman update
for state in states:
    action = "right"

    reward = R[(state, action)]
    next_s = next_state[(state, action)]

    # State-value function
    V[state] = reward + gamma * V[next_s]

    # Action-value function
    Q = reward + gamma * V[next_s]

    print("State:", state)
    print("Action:", action)
    print("V(s):", V[state])
    print("Q(s,a):", Q)
    print()
```

```python
#4. Exploration–Exploitation: ε-Greedy vs Softmax
import random
import math

q_values = [2, 5, 8]
actions = ["A", "B", "C"]

# Epsilon-Greedy
epsilon = 0.2

if random.random() < epsilon:
    action = random.choice(actions)
    print("Epsilon-Greedy: Exploring")
else:
    action = actions[q_values.index(max(q_values))]
    print("Epsilon-Greedy: Exploiting")

print("Selected Action:", action)


# Softmax
temperature = 1

exp_values = [math.exp(q / temperature) for q in q_values]
total = sum(exp_values)

probabilities = [x / total for x in exp_values]

print("\nSoftmax Probabilities:")
for i in range(len(actions)):
    print(actions[i], ":", round(probabilities[i], 3))

action = random.choices(actions, weights=probabilities)[0]

print("Softmax Selected Action:", action)
```

```python
#5. Reward Shaping
# Reward Shaping Example

position = 0
goal = 5

total_reward = 0

while position < goal:

    position += 1

    # Shaped reward
    if position == goal:
        reward = 10
    else:
        reward = 1

    total_reward += reward

    print("Position:", position,
          "Reward:", reward)

print("Total Reward:", total_reward)
```

```python
#6. Challenges of RL in Real-World Robotics
# RL Challenges in Robotics

battery = 30
samples = 0
safe = True

for step in range(5):

    samples += 1
    battery -= 5

    # Safety check
    if battery < 10:
        safe = False
        print("Safety problem: Battery too low")
        break

    # Environment variability
    if step == 2:
        print("Environment changed: Obstacle detected")
    else:
        print("Normal environment")

print("\nSamples Used:", samples)
print("Battery Remaining:", battery)
print("Safe:", safe)
```

```python
#7. Influence of Discount Factor γ
# Effect of Discount Factor

rewards = [5, 5, 20]

gammas = [0.1, 0.5, 0.9]

for gamma in gammas:

    value = 0

    for reward in rewards:
        value = reward + gamma * value

    print("Gamma:", gamma)
    print("Value:", round(value, 2))
    print()
```

```python
#8. Model-Free RL vs Model-Based RL
# Model-Free vs Model-Based RL

states = ["A", "B", "C"]

# Model-based: Agent knows the environment model
transition = {
    "A": "B",
    "B": "C",
    "C": "A"
}

# Model-free: Agent learns values from experience
value = {
    "A": 0,
    "B": 0,
    "C": 0
}

rewards = {
    "A": 5,
    "B": 10,
    "C": 2
}

print("MODEL-BASED RL")

state = "A"

for i in range(3):
    next_state = transition[state]
    print(state, "->", next_state)
    state = next_state


print("\nMODEL-FREE RL")

state = "A"

for i in range(5):

    reward = rewards[state]

    # Learn value from experience
    value[state] = value[state] + 0.5 * (reward - value[state])

    print("State:", state,
          "Reward:", reward,
          "Learned Value:", round(value[state], 2))

    state = transition[state]
```

```python
#9. Q-Learning Update and Convergence
# Simple Q-Learning

states = [0, 1, 2]
actions = ["left", "right"]

Q = {
    0: {"left": 0, "right": 0},
    1: {"left": 0, "right": 0},
    2: {"left": 0, "right": 0}
}

alpha = 0.5
gamma = 0.9

for episode in range(10):

    state = 0

    while state != 2:

        action = "right"

        if state == 0:
            next_state = 1
            reward = 0
        else:
            next_state = 2
            reward = 10

        old_value = Q[state][action]

        best_next = max(Q[next_state].values())

        # Q-learning update
        Q[state][action] = old_value + alpha * (
            reward + gamma * best_next - old_value
        )

        state = next_state

    print("Episode:", episode + 1,
          "Q(0,right):", round(Q[0]["right"], 2),
          "Q(1,right):", round(Q[1]["right"], 2))

print("\nFinal Q-Table:")
print(Q)
```

```python
#10. On-Policy vs Off-Policy Learning
# On-Policy vs Off-Policy

alpha = 0.5
gamma = 0.9

# Initial Q-values
Q_sarsa = {
    "A": {"left": 0, "right": 0},
    "B": {"left": 0, "right": 0}
}

Q_qlearning = {
    "A": {"left": 0, "right": 0},
    "B": {"left": 0, "right": 0}
}

state = "A"
action = "right"

next_state = "B"
next_action = "left"

reward = 5

# SARSA - On Policy
Q_sarsa[state][action] = Q_sarsa[state][action] + alpha * (
    reward + gamma * Q_sarsa[next_state][next_action]
    - Q_sarsa[state][action]
)

# Q-Learning - Off Policy
best_next = max(Q_qlearning[next_state].values())

Q_qlearning[state][action] = Q_qlearning[state][action] + alpha * (
    reward + gamma * best_next
    - Q_qlearning[state][action]
)

print("SARSA Q-value:")
print(Q_sarsa)

print("\nQ-Learning Q-value:")
print(Q_qlearning)
```
